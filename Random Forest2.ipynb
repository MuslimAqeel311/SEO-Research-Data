```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error
```

```python
dataset=pd.read_csv(r'D:seo_dataset170.csv')
dataset
```

```python
columns_to_select = [
    'SSL Certificate', 'Keyword in title', 'Keyword in Description', 
    'Keywords in phrases ', 'Long tail-keyword', 'Latent Semantic indexing Keywords', 
    'Mobile Optimized', 'Outbound link quality', 'Outbound link theme', 
    'Sitemap', 'Multimedia', 'Image optimized', 'Duplicate Content', 
    'Relevant Content', 'Page Loading Speed via HTML', 'Rel=canonical', 
    'Grammar Spelling', 'Broken URL', 'Keyword in h1 tag', 'W3C Validator', 
    'User-Friendly Layout', 'Parked Domains', 'Content Provides value and unique', 
    'Site Architecture', 'Breadcrumb navigation', 'Youtube', 'Link from Competitor', 
    'Redirect Page', 'Page Level Relevancy', 'Link from Hub Pages', 
    'Core web vitals', 'Syncytial Content', 'Site updates', 
    'Quality of Internal Link', 'Link of bad Neighborhood', 'Big brand preference', 
    'Amp version', 'Contact Us Page', 'Page Speed via Chrome', 'Schema markup'
]

X = dataset[columns_to_select]
y = dataset['Ranking']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state =40)
model=RandomForestRegressor(n_estimators=40, random_state=40)
model = model.fit(X_train, y_train)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
y_pred=model.predict(X_test)
```

```python
print(model.score(X_train,y_train))
```

```python
mse=mean_squared_error(y_test, y_pred)
print(f'Mean Squared Error:{mse}')
```
